**purpose**: reproduce panel f of Supplementary Fig. 2: first 4 seconds of the raw (unfiltered) traces of the first recording site of each cortical layer of the anesthetized rat dense probe recording of Horvath et al. (`dataset/00_raw/recording_horvath_probe1` and `probe2`, 20 kHz). Source data and svg are saved in figures/8_source_data/supp_fig2f/.

**method (as in the paper's notebook supp_fig2_AE_validation_sws.ipynb)**: the site is the first one (in channel order) of the layer (layers 2 and 3 merged); layer 1 is read from probe 1 and layers 2/3 to 6 from probe 2, because probe 2 does not cover layer 1; the y-axis spans the minimum and maximum of the 4 seconds.

```python
# SETUP PACKAGES
import os
import numpy as np
import pandas as pd
import warnings
from matplotlib import pyplot as plt
import spikeinterface as si
warnings.simplefilter(action="ignore")

# SET PATHS
PROJ_PATH = "/home/steeve/steeve/epfl/code/spikebias/"
os.chdir(PROJ_PATH)
RECORDING_PATHS = {"L1": "dataset/00_raw/recording_horvath_probe1", "L2_3": "dataset/00_raw/recording_horvath_probe2", "L4": "dataset/00_raw/recording_horvath_probe2",
                   "L5": "dataset/00_raw/recording_horvath_probe2", "L6": "dataset/00_raw/recording_horvath_probe2"}

# PARAMETERS
SFREQ = 20000                   # sampling frequency (Hz)
DUR_SEC = 4                     # duration (seconds)
LAYERS = ["L1", "L2_3", "L4", "L5", "L6"]
LAYER_LABELS = ["L1", "L2/3", "L4", "L5", "L6"]
LAYER_COLORS = [(0.7137, 0.6157, 0.3765), (0.7608, 0.4549, 0.5843), (0.6431, 0.5333, 0.6627), (0.3294, 0.4274, 0.4980), (0.5647, 0.7294, 0.5098)]
TRACE_COLOR = (0.3, 0.3, 0.3)

# FIGURE SETTINGS
plt.rcParams["font.family"] = "Arial"
plt.rcParams["font.size"] = 6  # 5-7 with Nature neuroscience as reference
plt.rcParams["lines.linewidth"] = 0.3
plt.rcParams["axes.linewidth"] = 1
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
plt.rcParams["xtick.major.width"] = 1
plt.rcParams["ytick.major.width"] = 1
plt.rcParams["xtick.major.size"] = 3.5
plt.rcParams["ytick.major.size"] = 3.5
savefig_cfg = {"transparent": True, "dpi": 300}
```

## Load the first 4 seconds of the first site of each layer

```python
%%time

df = pd.DataFrame({"time_s": np.arange(DUR_SEC * SFREQ) / SFREQ})
site_ids = {}
for layer in LAYERS:
    Rec = si.load_extractor(RECORDING_PATHS[layer]).frame_slice(0, DUR_SEC * SFREQ)       # lazy: only 4 seconds are read
    layer_of_site = np.array(["L2_3" if x in ("L2", "L3") else x for x in Rec.get_property("layers")])
    site_id = Rec.channel_ids[layer_of_site == layer][0]      # first site of the layer
    site_ids[layer] = site_id
    df[f"{layer}_voltage_uV"] = Rec.get_traces(channel_ids=[site_id])[:, 0]
print("sites:", site_ids)
display(df.describe().loc[["min", "max"]].round(1))
```
