# Supplementary Fig. 3b: spike shapes of inhibitory model neurons recorded above and below the somata

**author**: laquitainesteeve@gmail.com

**purpose**: reproduce panel b of Supplementary Fig. 3: spike-triggered average of the simulated extracellular voltage of 20 inhibitory neurons within 50 um of the Neuropixels probe, one per morphological type, recorded by an electrode 20 um above (blue) and below (purple) the soma. Source data and svg are saved in figures/8_source_data/supp_fig3b/.

**data**: `dataset/01_intermediate/supplementary/spike_shapes/spike_shapes_above_below_somata.csv`. The simulation output that produced the waveforms (bluepy circuit simulation) is not in the repository, so the values were reconstructed from the vector figure that the original notebook generated by `src/pipes/validation/spike_shapes/extract_spike_shapes_from_svg.py` (see the README in that folder for the precision of the reconstruction).

```python
# SETUP PACKAGES
import os
import numpy as np
import pandas as pd
import warnings
from matplotlib import pyplot as plt
warnings.simplefilter(action="ignore")

# SET PATHS
PROJ_PATH = "/home/steeve/steeve/epfl/code/spikebias/"
os.chdir(PROJ_PATH)
DATA_PATH = "dataset/01_intermediate/supplementary/spike_shapes/spike_shapes_above_below_somata.csv"

# PARAMETERS
COLORS = {"above": "tab:blue", "below": "tab:purple"}

# FIGURE SETTINGS
plt.rcParams["font.size"] = 6  # 5-7 with Nature neuroscience as reference
plt.rcParams["lines.linewidth"] = 0.5
plt.rcParams["axes.linewidth"] = 0.5
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
plt.rcParams["xtick.major.width"] = 0.5
plt.rcParams["ytick.major.width"] = 0.5
plt.rcParams["xtick.major.size"] = 3.5 * 1.1
plt.rcParams["ytick.major.size"] = 3.5 * 1.1
savefig_cfg = {"transparent": True, "dpi": 300}
```

## Load the spike shapes

```python
df = pd.read_csv(DATA_PATH)
df = df[df["panel"] == "b"].drop(columns="panel").reset_index(drop=True)
print(df["mtype"].nunique(), "inhibitory neurons")
df.head(3)
```

## Plot, same as the paper's panel a

```python
def plot_source_data(df: pd.DataFrame):
    """Spike waveforms (mV) of one inhibitory neuron per morphological type recorded 20 um above (blue) and 20 um below (purple) the soma"""

    n_rows, n_cols = df["row"].max(), df["column"].max()
    fig, ax = plt.subplots(n_rows, n_cols, figsize=(3, 7))
    for (row, col), d_cell in df.groupby(["row", "column"]):
        a = ax[row - 1, col - 1]
        for electrode in ["above", "below"]:
            d = d_cell[d_cell["electrode"] == electrode]
            a.plot(d["time_ms"], d["voltage_mV"], color=COLORS[electrode])
        a.text(0.98, 0.02, d_cell["mtype"].iloc[0], transform=a.transAxes, ha="right", va="bottom", fontsize=7)

        # disconnect axes (R style)
        a.spines["bottom"].set_position(("axes", -0.05))
        a.yaxis.set_ticks_position("left")
        a.spines["left"].set_position(("axes", -0.05))
        a.set_xlim([-1.2, 3.2])
        a.set_xticks([])
        if row == n_rows:
            a.set_xticks([-1, 0, 3], [-1, 0, 3])
            a.set_xlabel("Time (ms)")


    # legend
    fig.suptitle("Inhibitory model neurons", fontsize=7, fontweight="bold", y=1.0)
    for i, (label, color) in enumerate([("Above somata", "tab:blue"), ("Below somata", "tab:purple")]):
        fig.text(0.45, 0.975 - i * 0.012, label, color="k", fontsize=7, va="center")
        fig.add_artist(plt.Line2D([0.33, 0.43], [0.975 - i * 0.012] * 2, color=color, linewidth=2))
    fig.tight_layout(pad=0.5, rect=(0, 0, 1, 0.96))
    return fig
```

## Plot and save the figure and the source data

```python
fig = plot_source_data(df)
os.makedirs("figures/8_source_data/supp_fig3b", exist_ok=True)
plt.savefig("figures/8_source_data/supp_fig3b/supp_fig3b.svg", bbox_inches="tight", **savefig_cfg)
df.to_excel("figures/8_source_data/supp_fig3b/supp_fig3b.xlsx", index=False)
```
