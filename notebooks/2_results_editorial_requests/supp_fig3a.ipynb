## Load the spike shapes

```python
df = pd.read_csv(DATA_PATH)
df = df[df["panel"] == "a"].drop(columns="panel").reset_index(drop=True)
print(df["mtype"].nunique(), "excitatory neurons")
df.head(3)
```

## Plot, same as the paper's panel a

```python
def plot_source_data(df: pd.DataFrame):
    """Spike waveforms (mV) of one excitatory neuron per morphological type recorded 20 um above (red) and 20 um below (orange) the soma"""

    n_rows, n_cols = df["row"].max(), df["column"].max()
    fig, ax = plt.subplots(n_rows, n_cols, figsize=(3, 7))
    for (row, col), d_cell in df.groupby(["row", "column"]):
        a = ax[row - 1, col - 1]
        for electrode in ["above", "below"]:
            d = d_cell[d_cell["electrode"] == electrode]
            a.plot(d["time_ms"], d["voltage_mV"], color=COLORS[electrode])
        a.text(0.98, 0.02, d_cell["mtype"].iloc[0], transform=a.transAxes, ha="right", va="bottom", fontsize=7)

        # disconnect axes (R style)
        a.spines["bottom"].set_position(("axes", -0.05))
        a.yaxis.set_ticks_position("left")
        a.spines["left"].set_position(("axes", -0.05))
        a.set_xlim([-1.2, 3.2])
        a.set_xticks([])
        if row == n_rows:
            a.set_xticks([-1, 0, 3], [-1, 0, 3])
            a.set_xlabel("Time (ms)")
    ax[n_rows // 2, 0].set_ylabel("Voltage (mV)")

    # annotations of the paper (spike types)
    ax[0, 0].text(0.35, 0.95, "Positive spikes", transform=ax[0, 0].transAxes, color="tab:red", fontsize=5)
    ax[0, 0].text(0.55, 0.62, "Regular spike", transform=ax[0, 0].transAxes, color="tab:orange", fontsize=5)
    ax[5, 0].text(0.25, 0.97, "Triphasic spike", transform=ax[5, 0].transAxes, color="tab:red", fontsize=5)
    ax[5, 0].text(0.25, 0.02, "Regular spike", transform=ax[5, 0].transAxes, color="tab:orange", fontsize=5)

    # legend
    fig.suptitle("Excitatory model neurons", fontsize=7, fontweight="bold", y=1.0)
    for i, (label, color) in enumerate([("Above somata", "tab:red"), ("Below somata", "tab:orange")]):
        fig.text(0.45, 0.975 - i * 0.012, label, color="k", fontsize=7, va="center")
        fig.add_artist(plt.Line2D([0.33, 0.43], [0.975 - i * 0.012] * 2, color=color, linewidth=2))
    fig.tight_layout(pad=0.5, rect=(0, 0, 1, 0.96))
    return fig
```

## Plot and save the figure and the source data

```python
fig = plot_source_data(df)
os.makedirs("figures/8_source_data/supp_fig3a", exist_ok=True)
plt.savefig("figures/8_source_data/supp_fig3a/supp_fig3a.svg", bbox_inches="tight", **savefig_cfg)
df.to_excel("figures/8_source_data/supp_fig3a/supp_fig3a.xlsx", index=False)
```
