**author**: laquitainesteeve@gmail.com

**purpose**: reproduce panel e of Supplementary Fig. 2: first 4 seconds of the raw (unfiltered) traces of the first recording site of each cortical layer of the Neuropixels simulation of the evoked regime (`dataset/00_raw/recording_npx_evoked`, 20 kHz). Source data and svg are saved in figures/8_source_data/supp_fig2e/.

**method (as in the paper's notebook supp_fig2_AE_validation_sws.ipynb)**: the site is the first one (in channel order) of the layer (layers 2 and 3 merged); the y-axis spans the minimum (rounded down) and maximum (rounded up) of the 4 seconds, as in the paper's code.

```python
# SETUP PACKAGES
import os
import numpy as np
import pandas as pd
import warnings
from matplotlib import pyplot as plt
import spikeinterface as si
warnings.simplefilter(action="ignore")

# SET PATHS
PROJ_PATH = "/home/steeve/steeve/epfl/code/spikebias/"
os.chdir(PROJ_PATH)
RECORDING_PATH = "dataset/00_raw/recording_npx_evoked"

# PARAMETERS
SFREQ = 20000                   # sampling frequency (Hz)
DUR_SEC = 4                     # duration (seconds)
LAYERS = ["L1", "L2_3", "L4", "L5", "L6"]
LAYER_LABELS = ["L1", "L2/3", "L4", "L5", "L6"]
LAYER_COLORS = [(0.7137, 0.6157, 0.3765), (0.7608, 0.4549, 0.5843), (0.6431, 0.5333, 0.6627), (0.3294, 0.4274, 0.4980), (0.5647, 0.7294, 0.5098)]
TRACE_COLOR = (0.3, 0.3, 0.3)

# FIGURE SETTINGS
plt.rcParams["font.family"] = "Arial"
plt.rcParams["font.size"] = 6  # 5-7 with Nature neuroscience as reference
plt.rcParams["lines.linewidth"] = 0.3
plt.rcParams["axes.linewidth"] = 1
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
plt.rcParams["xtick.major.width"] = 1
plt.rcParams["ytick.major.width"] = 1
plt.rcParams["xtick.major.size"] = 3.5
plt.rcParams["ytick.major.size"] = 3.5
savefig_cfg = {"transparent": True, "dpi": 300}
```

## Load the first 4 seconds of the first site of each layer

```python
%%time

Rec = si.load_extractor(RECORDING_PATH).frame_slice(0, DUR_SEC * SFREQ)       # lazy: only 4 seconds are read
layer_of_site = np.array(["L2_3" if x in ("L2", "L3") else x for x in Rec.get_property("layers")])

df = pd.DataFrame({"time_s": np.arange(DUR_SEC * SFREQ) / SFREQ})
site_ids = {}
for layer in LAYERS:
    site_id = Rec.channel_ids[layer_of_site == layer][0]      # first site of the layer
    site_ids[layer] = site_id
    df[f"{layer}_voltage_uV"] = Rec.get_traces(channel_ids=[site_id])[:, 0]
print("sites:", site_ids)
display(df.describe().loc[["min", "max"]].round(1))
```

## Plot, same as the paper's panel c

```python
def plot_source_data(df: pd.DataFrame):
    """Voltage traces (microvolts) of the first 4 seconds of one site per layer, the y-axis spans the minimum and maximum of each trace"""

    fig, ax = plt.subplots(len(LAYERS), 1, figsize=(2.3, 2.5))
    time = df["time_s"].values
    for i, layer in enumerate(LAYERS):
        trace = df[f"{layer}_voltage_uV"].values
        ax[i].plot(time, trace, color=TRACE_COLOR, rasterized=True)

        # disconnect axes (R style)
        ax[i].spines["bottom"].set_position(("axes", -0.05))
        if i != len(LAYERS) - 1:
            ax[i].spines["bottom"].set_visible(False)
        ax[i].yaxis.set_ticks_position("left")
        ax[i].spines["left"].set_position(("axes", -0.01))
        # x
        ax[i].set_xlim([0, DUR_SEC])
        ax[i].set_xticks([])
        if i == len(LAYERS) - 1:
            ax[i].set_xticks([0, DUR_SEC], [0, DUR_SEC])
            ax[i].set_xlabel("Time (seconds)")
        # y
        lo, hi = np.floor(trace.min()), np.ceil(trace.max())
        ax[i].set_yticks([lo, hi], [f"{lo:.1f}", f"{hi:.1f}"])
        ax[i].set_ylim([lo, hi])
        ax[i].tick_params(direction="in")
        ax[i].text(1.0, 1.0, LAYER_LABELS[i], transform=ax[i].transAxes, color=LAYER_COLORS[i], fontsize=6, fontweight="bold", ha="right", va="bottom")
    fig.suptitle("Biophy. model of the evoked regime", fontsize=6, fontweight="bold", y=0.98)
    fig.tight_layout(pad=0.1, h_pad=0)
    return fig
```

## Plot and save the figure and the source data

```python
%%time

fig = plot_source_data(df)
os.makedirs("figures/8_source_data/supp_fig2e", exist_ok=True)
plt.savefig("figures/8_source_data/supp_fig2e/supp_fig2e.svg", **savefig_cfg)
df.to_excel("figures/8_source_data/supp_fig2e/supp_fig2e.xlsx", index=False)
```
